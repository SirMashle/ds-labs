```python
import math


def is_prime(n: int) -> bool:
    """Проверяет, является ли целое число n простым.

    Аргументы:
        n (int): Проверяемое число (n > 1 по условию).

    Возвращает:
        bool: True, если число простое, False - если составное или <= 1.
    """
    if n <= 1:
        return False

    limit: int = math.isqrt(n)

    for divisor in range(2, limit + 1):
        if n % divisor == 0:
            return False  # Найден делитель, число составное

    return True  # Делители не найдены, число простое
```

```python
# Тестовые примеры из условия
test_numbers = [29, 25, 2, 17, 49, 97, 100]

print("Демо:")
for num in test_numbers:
    result = is_prime(num)
    status = "простое" if result else "составное"
    print(f"Число {num:>3} -> {result!s:<5} ({status})")
```
